```python
import os, glob, time
from google.colab import drive
drive.mount('/content/drive')
MY = '/content/drive/MyDrive'

def locate(known, pattern, what):
    """Check the expected locations first; only walk the Drive if they miss."""
    for k in known:
        if os.path.exists(k):
            return k
    print(f'  {what}: not at the expected path, searching Drive...')
    hits = glob.glob(pattern, recursive=True)
    if not hits:
        raise SystemExit(f'Could not find {what}. Looked for: {pattern}')
    hits.sort(key=len)
    return hits[0]

_pipe = locate([f'{MY}/ber_code/business_entity_resolution/src/pipeline.py'],
               f'{MY}/**/business_entity_resolution/src/pipeline.py', 'code folder')
CODE_DRIVE = os.path.dirname(os.path.dirname(_pipe))

_train = locate([f'{MY}/student_resource/dataset/train/train_source1.tsv'],
                f'{MY}/**/train/train_source1.tsv', 'dataset')
DATA_DRIVE = os.path.dirname(os.path.dirname(_train))

VALIDATOR = locate([f'{MY}/student_resource/utils/validate_submission.py'],
                   f'{MY}/**/validate_submission.py', 'validator')

print('code on Drive:', CODE_DRIVE)
print('data on Drive:', DATA_DRIVE)
print('validator    :', VALIDATOR)
```
